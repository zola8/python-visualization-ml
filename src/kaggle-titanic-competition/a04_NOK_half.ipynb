# Load data

```python
import pandas as pd

train_data = pd.read_csv('train.csv')
test_data = pd.read_csv('test.csv')
```

```python
from sklearn.model_selection import train_test_split

y = train_data.Survived
X = train_data[['PassengerId', 'Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Embarked']]

# Divide data into training and validation subsets
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, test_size=0.2, random_state=1)

X_train.head()
```

# Handle missing values

```python
X_train.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer

# Imputation
mean_imputer = SimpleImputer(strategy="mean")
X_train["Age"] = mean_imputer.fit_transform(X_train[["Age"]])
# X_test["Age"] = mean_imputer.transform(X_test[["Age"]])
X_test["Age"] = mean_imputer.fit_transform(X_test[["Age"]])

# mean_imputer = SimpleImputer(strategy='most_frequent')# strategy can also be mean or median
# train_df.iloc[:,:] = mean_imputer.fit_transform(train_df)
# test_df.iloc[:,:] = mean_imputer.fit_transform(test_df)

X_test
```

```python
print("Duplicate values in training data is: ",X_train.duplicated().sum())
print("Duplicate values in testing data is: ",X_test.duplicated().sum())
```

# Handle categorical data

```python
X_train[:3]
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Handling categorical data - ordinal encoder is enough, we would do the same with one hot encoder too
ordinal_encoder = OrdinalEncoder()
object_cols = ["Sex", "Embarked"]

X_train[object_cols] = ordinal_encoder.fit_transform(X_train[object_cols])
X_test[object_cols] = ordinal_encoder.transform(X_test[object_cols])
```
