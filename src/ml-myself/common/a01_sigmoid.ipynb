# Logistic Regression in Machine Learning

https://www.geeksforgeeks.org/machine-learning/understanding-logistic-regression/

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 200)
z = 1/(1 + np.exp(-x))

plt.plot(x, z)
plt.xlabel("x")
plt.ylabel("Sigmoid(X)")

plt.show()
```
