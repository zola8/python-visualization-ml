```python
import seaborn as sns
```

```python
def visualization_scatterplot():
    sns.set_theme(style="darkgrid")
    sns.scatterplot(x=[1,2,3], y=[1,2,3])
```

```python
visualization_scatterplot()
```

```python
tips = sns.load_dataset("tips")

tips
```

```python
sns.relplot(
    data=tips,
    x="total_bill", y="tip", col="time",
    hue="smoker", style="smoker", size="size",
)
```

```python
sns.scatterplot(tips, x='total_bill', y='tip')
```

```python
sns.scatterplot(tips, x='total_bill', y='tip', hue='sex')
```

```python
sns.scatterplot(tips, x='total_bill', y='tip', hue='sex', size='size', style='smoker')
```

```python
sns.relplot(tips, x='total_bill', y='tip', hue='sex', size='size', style='smoker', kind='scatter', col='day', col_wrap=2)
```
