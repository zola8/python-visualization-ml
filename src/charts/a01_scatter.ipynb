```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
X = np.linspace(-np.pi, np.pi, 100)
y = np.sin(X)
```

```python
plt.figure(figsize=(10, 8))
plt.scatter(X, y)
plt.xlabel('X values')
plt.ylabel('sin(x)')
plt.grid(axis = 'y', alpha = 0.3)
```

```python
colors = ['red', 'blue', 'green', 'yellow']

for i in range(4):
    plt.plot(X, np.sin(X + i), color = colors[i], alpha = 0.3)
```
