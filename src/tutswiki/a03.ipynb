```python
# which borough has the most noise complaints?

is_noise = complaints['Complaint Type'] == 'Noise - Street/Sidewalk'
complaints[is_noise][['Borough']].value_counts()
```

```python
noise_complaints_count = complaints[is_noise][['Borough']].value_counts()
compliant_counts = complaints['Borough'].count()
```

```python
(noise_complaints_count / compliant_counts.astype(float)) * 100
```

```python
complaints['Complaint Type'].value_counts()
```

```python
(noise_complaints_count / compliant_counts).plot(kind='bar', x='Borough', y='Complaint Type')
```
