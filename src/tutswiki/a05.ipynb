```python
import pandas as pd
```

```python
weather = pd.read_csv('weather_2012.csv')
weather[:10]
```

```python
weather[['Date/Time', 'Temp (C)']]
```

```python
weather[['Date/Time', 'Temp (C)']].plot(kind='line', figsize=(15,5))
```

```python
weather.columns
```

```python
temperatures = weather[['Date/Time', 'Temp (C)']].copy()

temperatures
```

```python
temperatures.set_index('Date/Time', inplace=True)
```

```python
temperatures
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(15,5))
plt.xlabel('Date/Time')
plt.ylabel('Temperature (°C)')
plt.title('Temperature Over Time')
plt.grid(True, alpha=0.3)
plt.plot(temperatures.index, temperatures['Temp (C)'])
```

```python
temperatures.plot(figsize=(15,5))
```

```python
temperatures.index = pd.to_datetime(temperatures.index)
```

```python
temperatures['hour'] = temperatures.index.hour
```

```python
temperatures
```

```python
temperatures.groupby('hour').mean()
```

```python
temperatures.groupby('hour').mean().plot(figsize=(15,5))
```
