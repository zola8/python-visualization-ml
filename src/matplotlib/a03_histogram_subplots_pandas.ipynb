# Histogram

Histograms show the distribution of numerical data by grouping values into bins.

```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline

plt.style.use('default')
np.random.seed(234)
```

```python
data = np.random.normal(100, 15, 1000)  # Mean=100, Std=15, 1000 points

plt.figure(figsize=(10, 6))
plt.hist(data, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Normal Distribution')
plt.grid(True, axis='y', alpha=0.3)
plt.show()
```

```python
data1 = np.random.normal(100, 15, 1000)  # Mean=100, Std=15, 1000 points
data2 = np.random.normal(120, 20, 1000)

plt.figure(figsize=(10, 6))
plt.hist(data1, bins=30, color='blue', alpha=0.6, label='Dataset 1')
plt.hist(data2, bins=30, color='red', alpha=0.6, label='Dataset 2')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Overlapping Histograms')
plt.grid(True, axis='y', alpha=0.3)
plt.show()
```

# Subplots

Subplots allow you to create multiple plots in a single figure, which is great for comparing different visualizations.

```python
x = np.linspace(0, 10, 100)

fig, axes = plt.subplots(2, 2, figsize=(10, 6))
fig.suptitle('Multiple Subplots', fontsize=16, fontweight='bold')

# top, left - line plot
axes[0, 0].plot(x, np.sin(x), 'b-', linewidth=2)
axes[0, 0].set_title('sin(x)')
axes[0, 0].grid(True, alpha=0.3)

# top right - line plot
axes[0, 1].plot(x, np.cos(x), 'r-', linewidth=2)
axes[0, 1].set_title('cos(x)')
axes[0, 1].grid(True, alpha=0.3)

# bottom left - scatter
x_scatter = np.random.randn(100)
y_scatter = np.random.randn(100)
axes[1, 0].scatter(x_scatter, y_scatter, alpha=0.6, c=x_scatter, cmap='viridis')
axes[1, 0].set_title('Scatter Plot')
axes[1, 0].grid(True, alpha=0.3)

# bottom right - bar chart
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 75, 37]
axes[1, 1].bar(categories, values, color='coral')
axes[1, 1].set_title('Bar Chart')
axes[1, 1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```

```python
# pandas
dates = pd.date_range(start='2026-01-01',periods=100, freq='D')
df = pd.DataFrame({
    'Date': dates,
    'Sales': np.random.randint(1000, 5000, 100) + np.sin(np.arange(100)) * 500,
    'Profit': np.random.randint(200, 800, 100) + np.cos(np.arange(100)) * 200,
    'Region': np.random.choice(['North', 'South', 'East', 'West'], 100),
})

print("Sample dataframe")
print(df.head())
print("...")
print(df.tail())
print(f"Shape: {df.shape}")
```

```python
plt.figure(figsize=(14, 6))
plt.plot(df['Date'], df['Sales'], 'b-', label='Sales')
plt.plot(df['Date'], df['Profit'], 'g-', label='Profit')
plt.xlabel('Date')
plt.ylabel('Amount ($)')
plt.legend()
plt.grid(True, axis='y', alpha=0.3)
plt.title('Sales vs Profit')
plt.xticks(rotation=45)
plt.show()
```
