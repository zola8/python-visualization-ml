```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
# only for jupyter nootbook to display plots inline
```

```python
plt.style.use('default')
```

```python
x = [1,2,3,4,5]
y = [2,4,7,8,10]

plt.plot(x,y)
plt.plot(y,x)

plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Simple line plot')

# in jupyter nootbook the plot is displayed automatically, it is useful when working with py files
plt.show()
```

```python
# subplots - creates a figure and axes object (more control for complex plots)
fig, ax = plt.subplots()

ax.plot(x,y)
ax.plot(y,x)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Simple line plot')
plt.show()
```

# Basic Line Plots

```python
# create sample data
x = np.linspace(0,10,100) # 100 points from 0 to 10
y1 = np.sin(x)
y2 = np.cos(x)
y3 = y1 * y2 + 1

plt.figure(figsize=(10,5)) # width, height in inches

plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)

plt.xlabel('X Axis')
plt.ylabel('sin/cos functions')
plt.title('sin(x) function')
plt.grid(True, alpha=0.3) # grid transparency
plt.legend(['sin(x)', 'cos(x)', 'sin(x)*cos(x)'])
```

# Customization of plots

Colors, Styles, Markers

```python
x = np.linspace(0,10,20)
plt.figure(figsize=(12,6))

plt.plot(x, x, 'b-', label='Blue solid line', linewidth=2)
plt.plot(x, x+1, 'g-', label='Green solid line', linewidth=2)
plt.plot(x, x+2, 'r--', label='Red dashed line', linewidth=2)
plt.plot(x, x+3, 'm-.', label='Magenta dash-dot line', linewidth=2)
plt.plot(x, x+4, 'c:', label='Cyan dotted line', linewidth=2)
plt.plot(x, x+5, 'co-', label='Circles', linewidth=2)
plt.plot(x, x+6, 's-', label='Squares', linewidth=2)
plt.plot(x, x+7, '^-', label='Triangles', linewidth=2)


plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Line styles')
plt.legend()
```
