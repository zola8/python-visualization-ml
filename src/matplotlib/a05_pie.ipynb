```python

import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

plt.style.use('default')
np.random.seed(234)
```

```python
categories = ['North', 'South', 'West', 'East']
sizes = [30, 25, 30, 40]
colors = ['#A23464', '#F63214', '#894AB1', '#365434']
explode = (0, 0.2, 0, 0)

fig, ax = plt.subplots(figsize=(8, 8))

ax.pie(sizes, explode=explode, labels=categories, colors=colors, shadow=False)
ax.set_title("Sales by region")
plt.show()

# save fig
fig.savefig("a05_pie.png", dpi=300, bbox_inches='tight')
fig.savefig("a05_pie.pdf")
```
