# Scatter Plots

https://www.youtube.com/watch?v=7Lc2AxiM17o


```python

import matplotlib.pyplot as plt
import numpy as np
```

```python
%matplotlib inline

plt.style.use('default')
```

A scatter plot represents the values of two different numeric variables, with one on the horizontal (x-axis) and one on the vertical (y-axis).

```python
np.random.seed(123)
x = np.random.randn(100)
y = np.random.randn(100)
colors = np.random.rand(100)
sizes = 1000 * np.random.rand(100)

plt.figure(figsize=(10, 6))
plt.scatter(x, y, c=colors, s=sizes, alpha=0.5)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.colorbar(label="Color incensity")
plt.show()

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='plasma')
plt.show()

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='Set1')
plt.show()
```

# Bar Charts

Ideal for comparing categorical data or showing counts.

```python
categories = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']
values = [23, 54, 37, 85, 45, 60, 11, 94]

# vertical bar chart

plt.figure(figsize=(10, 6))
plt.bar(categories, values, color='steelblue', edgecolor='black')
plt.xlabel("Categories")
plt.ylabel("Values")
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D', 'E', 'F']
values = [23, 54, 37, 85, 45, 60]

# horizontal bar chart
plt.figure(figsize=(10, 6))
plt.barh(categories, values, color='coral', edgecolor='black', linewidth=1)
plt.ylabel("Categories")
plt.xlabel("Values")
plt.grid(True, alpha=0.3, axis='x')
plt.show()
```

```python
# grouped bar chart

categories = ['Q1', 'Q2', 'Q3', 'Q4']
product_a = [20, 35, 30, 35]
product_b = [25, 32, 34, 20]
product_c = [30, 25, 30, 25]

x = np.arange(len(categories))
width = 0.25  # width of bar

plt.figure(figsize=(10, 6))
plt.bar(x - width, product_a, width, color='#B27456', edgecolor='black', linewidth=1, label='Product A')
plt.bar(x, product_b, width, color='#542256', edgecolor='black', linewidth=1, label='Product B')
plt.bar(x + width, product_c, width, color='#7425A7', edgecolor='black', linewidth=1, label='Product C')

plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.title("Quarterly Sales Comparison")
plt.grid(True, alpha=0.2, axis='y')
plt.xticks(x, categories)
plt.legend()
plt.show()
```
